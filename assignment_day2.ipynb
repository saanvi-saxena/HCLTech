```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
tips = px.data.tips()
iris = px.data.iris()
stocks = px.data.stocks()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 14, 20, 24, 22]
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales (in thousands)")
plt.grid(True)
plt.show()
```

```python
avg_tip = tips.groupby("day")["tip"].mean().reindex(
    ["Thur", "Fri", "Sat", "Sun"]
)
plt.bar(avg_tip.index, avg_tip.values, color="skyblue")
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(tips["total_bill"], bins=10, color="skyblue")
axes[0].set_title("Distribution of Total Bills")
axes[0].set_xlabel("Total Bill")
axes[0].set_ylabel("Frequency")
axes[1].scatter(tips["total_bill"], tips["tip"], color="green")
axes[1].set_title("Total Bill vs Tip")
axes[1].set_xlabel("Total Bill")
axes[1].set_ylabel("Tip")
plt.tight_layout()
plt.savefig("tips_plots.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=tips, x="day")
plt.title("Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Number of Customers")
plt.show()
plt.figure(figsize=(6, 4))
sns.barplot(data=tips, x="day", y="total_bill")
plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Distribution of Total Bills by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
plt.figure(figsize=(7, 5))
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Total Bills by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.legend(title="Sex")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=tips, x="total_bill", y="tip",
    hue="smoker", style="time"
)
plt.title("Total Bill vs Tip by Smoker Status and Time")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.pairplot(
    iris,
    hue="species",
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"]
)
plt.show()
```

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="day",
    size="size",
    hover_data=["sex", "time"],
    title="Bill vs Tip"
)
fig.update_layout(xaxis_title="Total Bill", yaxis_title="Tip")
fig.show()
```

```python
fig = px.box(
    tips, x="day", y="total_bill", color="sex",
    title="Total Bill Distribution by Day and Sex"
)
fig.update_layout(xaxis_title="Day", yaxis_title="Total Bill")
fig.show()
fig = px.histogram(
    tips, x="total_bill", color="time", nbins=15,
    title="Distribution of Total Bills by Meal Time"

)
fig.update_layout(xaxis_title="Total Bill", yaxis_title="Frequency")
fig.show()
```

```python
fig = px.line(
    stocks, x="date", y=["GOOG", "AAPL", "AMZN"],
    title="Stock Prices from 2018 to 2019"
)
fig.update_layout(xaxis_title="Date", yaxis_title="Relative Stock Price")
fig.update_xaxes(rangeslider_visible=True)
fig.show()
```
