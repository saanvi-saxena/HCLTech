```python
#Inspect the annual_salary_inr column
print(emp['annual_salary_inr'].head(10))
print(emp['annual_salary_inr'].dtype)
print(emp['annual_salary_inr'].describe())
```

```python
#Convert salary values into numeric values
emp['annual_salary_inr'] = emp['annual_salary_inr'].astype('string')

emp['annual_salary_inr'] = emp['annual_salary_inr'].str.replace(
    r'[₹,\s]', '', regex=True
)

emp['annual_salary_inr'] = pd.to_numeric(
    emp['annual_salary_inr'], errors='coerce'
)

print(emp['annual_salary_inr'].head(10))
```

```python
#Handle values such as 850k and unknown
salary = emp['annual_salary_inr'].astype('string').str.strip()

salary = salary.str.replace(r'[₹,\s]', '', regex=True)

salary = salary.replace({
    'unknown': pd.NA,
    'Unknown': pd.NA,
    'UNKNOWN': pd.NA,
    '': pd.NA
})

salary = salary.str.replace(
    r'(?i)^(\d+(?:\.\d+)?)k$',
    lambda m: str(float(m.group(1)) * 1000),
    regex=True
)

emp['annual_salary_inr'] = pd.to_numeric(
    salary, errors='coerce'
)

print(emp['annual_salary_inr'].head(10))
```

```python
# Identify negative salaries
print(emp.loc[emp['annual_salary_inr'] < 0,
              ['employee_id', 'employee_name', 'annual_salary_inr']])

# Identify zero salaries
print(emp.loc[emp['annual_salary_inr'] == 0,
              ['employee_id', 'employee_name', 'annual_salary_inr']])

# Identify unusually high salaries
Q1 = emp['annual_salary_inr'].quantile(0.25)
Q3 = emp['annual_salary_inr'].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

print("Unusually high salaries:")
print(emp.loc[emp['annual_salary_inr'] > upper_limit,
              ['employee_id', 'employee_name', 'annual_salary_inr']])
```

```python
#Calculate mean, median, minimum, and maximum salary
print("Mean salary:", emp['annual_salary_inr'].mean())
print("Median salary:", emp['annual_salary_inr'].median())
print("Minimum salary:", emp['annual_salary_inr'].min())
print("Maximum salary:", emp['annual_salary_inr'].max())
```

```python
#Calculate average salary for each department
avg_salary = emp.groupby('department')['annual_salary_inr'].mean()

print(avg_salary)
```
